```python
import json

# Load the dataset (replace with the path to your actual dataset if necessary)
file_path = '/Users/fadisultan/PycharmProjects/llm-open-domain-table-reasoner/open_wikitable/data/train.json'

with open(file_path, 'r') as f:
    data = json.load(f)

# Extract the question and answer dictionaries
question_id_data= data.get('question_id', {})
questions_data = data.get('question', {})
answers_data = data.get('answer', {})
original_table_id_data = data.get('original_table_id', {})
sql_data = data.get('sql', {})
hard_positive_idx_data = data.get('hard_positive_idx', {})
positive_idx_data= data.get('positive_idx', {})
negative_idx_data = data.get('negative_idx', {})
dataset_data = data.get('dataset', {})





def get_question_and_answer(question_number: str):
    """
    Fetch question and answer based on the question number.
    """
    question_id =question_id_data.get(question_number, 'Question not found')
    answer = answers_data.get(question_number, 'Answer not found')
    question = questions_data.get(question_number, 'Question not found')
    original_table_id=original_table_id_data.get(question_number, 'original_table_id not found')
    sql=sql_data.get(question_number, 'sql not found')
    hard_positive_idx=hard_positive_idx_data.get(question_number, 'hard_positive_idx not found')
    positive_idx=positive_idx_data.get(question_number, 'positive_idx not found')
    negative_idx=negative_idx_data.get(question_number, 'negative_idx not found')
    dataset=dataset_data.get(question_number, 'dataset not found')

    return question, answer,question_id,original_table_id,sql,hard_positive_idx,positive_idx,negative_idx,dataset


# Take the question number as input from the user
question_number = input("Enter the question number (e.g., '4102'): ")

# Get the question and answer for the provided question number
question, answer, question_id,original_table_id,sql,hard_positive_idx,positive_idx,negative_idx,dataset= get_question_and_answer(question_number)

# Display the result
print(f"\nQuestion: {question}")
print(f"Answer: {answer}",)
print(f"question_id: {question_id}")
print(f"original_table_id: {original_table_id}")
print(f"sql: {sql}")
print(f"hard_positive_idx: {hard_positive_idx}")
print(f"positive_idx: {positive_idx}")
print(f"negative_idx: {negative_idx}")
print(f"dataset: {dataset}")

```
